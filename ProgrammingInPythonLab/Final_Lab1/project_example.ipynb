# Project Example

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Diamond dataset from github
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/diamonds.csv"
diamonds = pd.read_csv(url)
diamonds.shape
```

```python
# Dataset overview
print("Dataset shape:", diamonds.shape)
print("Dataset columns:", diamonds.columns.tolist())
print("First 5 rows of the dataset: \n", diamonds.head())
#Datatypes and missing values
print("Data types: ", diamonds.dtypes)
print("Missing values: \n", diamonds.isnull().sum())
#Summary statistics
print("Summary statistics: \n", diamonds.describe())
```
