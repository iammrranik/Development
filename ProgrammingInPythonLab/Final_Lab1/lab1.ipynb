Data Frame

```python
students = {
    "name": ["John", "Bob", "Eve", "Eric", "Jane"],
    "quiz": [90, 92, 96, 91, 14],
    "mid": [91, 92, 94, 96, 12],
    "final": [88, 91, 93, 99, 13],
    "attendance": [9, 9, 8, 9, 5]
}

df = pd.DataFrame(students)
df
```

```python
# df.head() # shows the first 5 rows of the dataframe
df.head(3) # shows the first 3 rows of the dataframe
```

```python
df.tail() # shows the last 5 rows of the dataframe
df.tail(2) # shows the last 2 rows of the dataframe
```

```python
df.shape # shows the number of rows and columns in the dataframe
```

```python
df.columns # shows the column names of the dataframe
```

```python
df.dtypes # shows the data types of each column in the dataframe
```

```python
df.info()
```

```python
df.describe() # shows the summary statistics of the numerical columns in the dataframe
```

```python
df['total'] = df['quiz'] + df['mid'] + df['final'] + df['attendance'] # creates a new column 'total' which is the sum of quiz, mid, final, and attendance scores
df['average'] = df[['quiz', 'mid', 'final']].mean(axis=1) # creates a new column 'average' which is the mean of quiz, mid and final
df['passed'] = df['average'] >= 50 # creates a new column 'passed' which is True if the average score is greater than or equal to 50, and False otherwise
df
```

```python
df[['name', 'average', 'passed']] # shows only the name, total, average and passed columns of the dataframe
```

```python
df.sort_values(by='average', ascending=False) # sorts the dataframe by the average column in descending order
```

```python
df.to_csv('students.csv', index=False) # saves the dataframe to a csv file named 'students.csv' without the index column
```
