```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
titanic = pd.read_csv(url)
titanic.shape
```

```python
# data cleaning
duplicates = titanic.duplicated().sum()
print("Number of duplicate rows:", duplicates)
if duplicates > 0:
    titanic = titanic.drop_duplicates()
    print("Duplicates removed. New dataset shape:", titanic.shape)
```

```python
# Plot survive distribution based on age
plt.figure(figsize=(10, 6))
plt.title("Survival Distribution by Age")
mean_survival_age = titanic.groupby('age')['survived'].mean()
mean_survival_age.plot(kind='bar', color='skyblue')
plt.xlabel("Age")
plt.ylabel("Mean Survival Rate")
plt.xticks(rotation=0)
# plt.tight_layout()
plt.show()
```
