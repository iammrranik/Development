### Problem: Maximize f(x) = x<sup>2</sup> - 11x + 150 when x = 0 to 255

Representation or ecoding solution:
x can be represented using 8 bits binary number.

# Step 1: Initialize Population

```python
import random
def initPopulation(b=8, n =10): # n is the number of solutions, b is the length
    p = [random.choices([0,1],k = b) for i in range(n)]
    return p
```

# Step 2: Reproduction or Parent Selection

Fitness Calculation

```python
def binToDecimal(binList):
  decValue = 0 # decimel value
  power = len(binList)
  for digit in binList: # binary to decimel
    decValue += digit*2**(power-1)
    power -= 1
  return decValue

def func(x):
  return x**2 - 11*x + 150

def getFitness(p): # p is the whole population
    f = [] # to store fitness values
    for sol in p:
        d = binToDecimal(sol)
        v = func(d)
        if v< 0: f.append(0)
        else: f.append(v) # fitness function
    return f
```

calculate probability and select parents

```python
population = initPopulation(8,10)
print(population)
```

```python
fitness = getFitness(population)
print(fitness)
total_fitness = sum(fitness)
print(total_fitness)
prob = [f/total_fitness for f in fitness]
print(prob)

# Select parents based on probability
parents = random.choices(population, weights=prob, k=10)
```

# Step 3: Crossover

```python
# perform crossover: first two pairs at point 4 and last three pairs at point 6
def cross(p):
  num_cross = len(p)//2
  print(num_cross)
  offsprings = []
  for i in range(num_cross):
    # Set crossover point based on the pair index
    if i < 2: cp = 4
    else: cp = 6
    
    # Select the pair
    parent1 = p[2*i]
    parent2 = p[2*i + 1]
    
    # Perform crossover to create two offspring per pair
    off1 = parent1[:cp] + parent2[cp:]
    off2 = parent2[:cp] + parent1[cp:]
    
    offsprings.append(off1)
    offsprings.append(off2)
    
  return offsprings
```

```python
off = cross(parents)
off
```

# Step 4: Mutation

```python
# Perform mutation with >3% probability at point 2 and 7
# def mutate(offsp, pr = 0.03):
```

```python
# Perform mutation with >3% probability at point 2 and 7
def mutate(offsp, pr = 0.03):
  for sol in offsp:
    if random.random() < pr:
      # Mutate at point 2 (index 1)
      sol[1] = 1 - sol[1]
    if random.random() < pr:
      # Mutate at point 7 (index 6)
      sol[6] = 1 - sol[6]
  return offsp

mutated_off = mutate(off)
```

# Step 5: Select Survivor
The selected survivors (solutions) will be added to population

```python
#write the code for survivor
```

```python
def selectSurvivor(old_pop, offspring):
    total_pop = old_pop + offspring
    
    total_fitness = getFitness(total_pop)
    
    combined = []
    for i in range(len(total_pop)):
        combined.append([total_fitness[i], total_pop[i]])
    
    combined.sort(reverse=True)
    
    new_p = []
    for i in range(10): # population size is 10
        new_p.append(combined[i][1])
        
    return new_p

population = selectSurvivor(population, mutated_off)
print("New Population Size:", len(population))
print("Best Fitness in current generation:", max(getFitness(population)))
```
